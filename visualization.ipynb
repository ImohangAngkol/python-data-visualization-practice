# Python Data Visualization Practice

This notebook demonstrates basic data visualization using Python, Pandas, and Matplotlib.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Language": ["Python", "Java", "C++", "JavaScript"],
    "Students": [45, 30, 20, 35]
}

df = pd.DataFrame(data)

df
```

```python
plt.bar(df["Language"], df["Students"])

plt.title("Programming Language Popularity")
plt.xlabel("Programming Language")
plt.ylabel("Number of Students")

plt.show()
```
